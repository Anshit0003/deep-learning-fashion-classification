# Step 1: Import the Required Libraries

We will use TensorFlow/Keras to create the neural network and Matplotlib to display images.

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

print("TensorFlow version:", tf.__version__)
```

# Step 2: Load the Fashion Product Dataset

We will use the **Fashion MNIST** dataset.

It contains images of common fashion products such as shirts, shoes, bags and dresses.

The dataset downloads automatically, so no manual file upload is required.

```python
fashion_mnist = tf.keras.datasets.fashion_mnist

(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

print("Training images:", train_images.shape)
print("Testing images:", test_images.shape)
```

# Step 3: Define the Product Categories

The model uses numbers internally.

We will connect those numbers to understandable product names.

```python
class_names = [
    "T-shirt/Top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle Boot"
]

print(class_names)
```

# Step 4: Look at One Product Image

Before training the model, let us see what one image looks like.

```python
plt.imshow(train_images[0], cmap="gray")
plt.title("Product: " + class_names[train_labels[0]])
plt.axis("off")
plt.show()
```

## Business Interpretation

Each image is a piece of data.

For an e-commerce company:

**Input:** Product image  
**Expected Output:** Product category

The goal of the neural network is to learn patterns from many labelled product images.

# Step 5: View Several Product Images

```python
plt.figure(figsize=(10, 6))

for i in range(12):
    plt.subplot(3, 4, i + 1)
    plt.imshow(train_images[i], cmap="gray")
    plt.title(class_names[train_labels[i]])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

# Step 6: Prepare the Images

Image pixel values originally range from 0 to 255.

We convert them to values between 0 and 1.

You only need to understand that this makes the image data easier for the neural network to process.

```python
train_images = train_images / 255.0
test_images = test_images / 255.0

print("Images are ready for training.")
```

# Step 7: Create a Simple Neural Network

Our model has three main stages:

### Input
The product image enters the network.

### Hidden Layer
The network learns useful patterns.

### Output
The model selects one of the 10 product categories.

```python
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])

model.summary()
```

## Connect This to Theory

The model contains:

**Input → Hidden Layer → Output**

- `Flatten` prepares the image for the network.
- `Dense(64)` acts as a hidden layer.
- `ReLU` is an activation function.
- The final layer has 10 outputs because there are 10 product categories.
- `Softmax` helps produce category probabilities.

No mathematical calculation is required for this course.

# Step 8: Prepare the Model for Training

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model is ready for training.")
```

# Step 9: Train the Neural Network

An **epoch** means the model has gone through the training dataset once.

We will use only 3 epochs to keep the classroom demonstration quick.

```python
history = model.fit(
    train_images,
    train_labels,
    epochs=3,
    validation_split=0.1
)
```

# Step 10: Check Model Accuracy

Now we test the model on images it did not use for training.

```python
test_loss, test_accuracy = model.evaluate(test_images, test_labels, verbose=0)

print("Test Accuracy:", round(test_accuracy * 100, 2), "%")
```

## What Does Accuracy Mean Here?

If the model shows, for example, **87% accuracy**, it means that approximately 87 out of every 100 test images were classified correctly.

The exact result may vary slightly.

### Management Question

Would accuracy alone be enough before a company deploys this system?

Think about:

- Cost of incorrect classification
- Customer experience
- Human review
- Quality of training data

# Step 11: Ask the Model to Predict New Images

```python
predictions = model.predict(test_images, verbose=0)

predicted_class = np.argmax(predictions[0])
actual_class = test_labels[0]

print("Predicted Product:", class_names[predicted_class])
print("Actual Product:", class_names[actual_class])
```

# Step 12: Display the Prediction

```python
plt.imshow(test_images[0], cmap="gray")
plt.title(
    "Predicted: " + class_names[predicted_class] +
    "\nActual: " + class_names[actual_class]
)
plt.axis("off")
plt.show()
```

# Step 13: Test Another Product

Change the value of `image_number` to try another image.

For example:

`image_number = 25`

```python
image_number = 25

prediction = model.predict(
    np.expand_dims(test_images[image_number], axis=0),
    verbose=0
)

predicted_class = np.argmax(prediction)
actual_class = test_labels[image_number]

plt.imshow(test_images[image_number], cmap="gray")
plt.title(
    "Predicted: " + class_names[predicted_class] +
    "\nActual: " + class_names[actual_class]
)
plt.axis("off")
plt.show()
```
